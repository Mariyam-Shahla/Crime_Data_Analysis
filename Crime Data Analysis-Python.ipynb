```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt 
import seaborn as sns
df=pd.read_excel("project_dataset.xlsx")
df
```

```python
df.head()
```

```python
df.shape
```

```python
df.columns
```

```python
df.isnull().sum()
```

```python
num_cols=df.select_dtypes(include=np.number).columns
df[num_cols]=df[num_cols].fillna(df[num_cols].median())
```

```python
cat_cols=df.select_dtypes(include='object').columns
for col in cat_cols:
    df[col].fillna(df[col].mode()[0])
                    
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df=df.drop_duplicates()
```

```python
df.info()
```

```python
df.describe()
```

```python
yearly_crime=df.groupby('Year')["Cases_Reported"].sum()
yearly_crime
```

```python
crime_type_total=df.groupby("Crime_Type")["Cases_Reported"].sum().sort_values(ascending=False)
crime_type_total
```

```python
df.loc[:,"Solving_Rate"]=(df["Cases_Solved"]/df["Cases_Reported"])*100
df
```

```python
df.loc[:, "Total_Victims"]=df["Female_Victims"]+df["Male_Victims"]
df[["Male_Victims","Female_Victims","Total_Victims"]].head()
```

```python
df.loc[:, "Victim_Burden_Index"]=(df["Total_Victims"]/df["Cases_Reported"])
df.head()
```

```python
df.loc[:, "Arrest_Rate"]=df["Arrests_Made"]/df["Cases_Reported"].replace(0,1)
df.head()
```

```python
df.loc[:, "Urban_Crime_Ratio"]=df["Urban_Cases"]/df["Cases_Reported"].replace(0,1)
df.head()
```

```python
df.loc[:, "Rural_Crime_Ratio"]=df["Rural_Cases"]/df["Cases_Reported"].replace(0,1)
df.head()
```

```python
df.shape
```

```python
area_data=df[["Urban_Cases","Rural_Cases"]].sum()
plt.figure(figsize=(6,6))
area_data.plot(kind="pie",autopct="%1.1f%%",startangle=90)
plt.title("Urban vs Rural Crime Distribution")
plt.ylabel(" ")
plt.show()
```

```python
print("Urban Cases are  comparatively higher than Rural Cases")
```

```python
numeric_df=df.select_dtypes(include=np.number)
corr_matrix=numeric_df.corr()

plt.figure(figsize=(10,6))
sns.heatmap(corr_matrix,cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
          
```

```python
print("The Heatmap shows a strong positive correlation between total crime cases,arrests and violent or women-related crimes.Solving rate and Conviction rate are moderately related to cases solved,while time to solve cases shows weaker correlation.") 
```

```python
bins=[0,12,18,30,45,60,100]
labels=["Child","Teen","Young Adult","Adult","Middle Age","Senior"]
df.loc[:, "Victim_Age_Group"]=pd.cut(df["Avg_Victim_Age"],bins=bins,labels=labels)
```

```python
df.loc[:, "Accused_Age_Group"]=pd.cut(df["Avg_Accused_Age"],bins=bins,labels=labels)
```

```python
df.groupby("Victim_Age_Group",observed=True)["Cases_Reported"].sum().plot(kind="bar",figsize=(8,5),title="Crimes by Victim Age Group")
plt.xlabel("Victim Age Group")
plt.ylabel("Total Cases")
plt.show()
```

```python
print("Most crimes are reported against adults and  young adults, while teenagers account for a much smaller share.This indicates that the working-age population is the most affected group by crime.")
```

```python
df.groupby("Accused_Age_Group",observed=True)["Cases_Reported"].sum().plot(kind="bar",color="orange",figsize=(8,5))
plt.title("Crimes by Accused Age Group")
plt.xlabel("Accused Age Group")
plt.ylabel("Cases Reported")
plt.show()
           
```

```python
print("Most crimes are committed by adults,followed by young adults,while middle-aged individuals are involved in comparatively fewer cases.This indicates that crime involvement is highest among the active working-age population.")
```
